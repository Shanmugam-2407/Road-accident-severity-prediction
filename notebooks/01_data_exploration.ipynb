# ROADGUARD AI: 01. Exploratory Data Analysis
This notebook analyzes historical road accident distributions, missingness patterns, and bivariate relationships with accident severity.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.utils.config import DATASET_PATH, TARGET_COLUMN

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv(DATASET_PATH)
print(f"Loaded {len(df)} accident records.")
df.head()
```

```python
print("Summary Information:")
df.info()
```

```python
print("Class Distribution:")
print(df[TARGET_COLUMN].value_counts(normalize=True))
```

```python
# Plot target severity imbalance
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x=TARGET_COLUMN, palette="Blues_r")
plt.title("Historical Accident Severity Distribution")
plt.xlabel("Severity Category")
plt.ylabel("Count")
plt.show()
```

```python
# Relationship with speed limits
plt.figure(figsize=(9, 4))
sns.countplot(data=df, x="speed_limit", hue=TARGET_COLUMN, palette="Blues_r")
plt.title("Accident Severity by Speed Limit Zone")
plt.show()
```
